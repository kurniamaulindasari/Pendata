#### METODE 3: PYTHON SCRIPT
Python Script merupakan node fleksibel di KNIME yang memungkinkan pengguna menulis kode Python untuk menerapkan algoritma machine learning lain yang tidak tersedia di node standar KNIME.

image.png



```python
import pandas as pd
import knime.scripting.io as knio
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# 1. BACA DATA DARI KNIME
df = knio.input_tables[0].to_pandas()

print("="*60)
print("ANALISIS RANDOM FOREST - DATA GRADE")
print("="*60)
print(f"Total data: {len(df)} baris")

# 2. HAPUS DATA KOSONG
df = df.dropna()
print(f"Data setelah hapus missing: {len(df)} baris")

# 3. TENTUKAN KOLOM TARGET DAN FITUR
target_col = 'GRADE'

# Ambil semua fitur kecuali GRADE, COURSE ID, STUDENT ID
fitur_cols = [col for col in df.columns if col not in [target_col, 'COURSE ID', 'STUDENT ID', 'S1=STUDENT ID', 'RowID']]

print(f"\nTarget: {target_col}")
print(f"Jumlah fitur: {len(fitur_cols)}")

# 4. PISAHKAN FITUR DAN TARGET
X = df[fitur_cols]
y = df[target_col].astype(str)

print(f"Kelas target: {y.unique().tolist()}")
print("Jumlah data per kelas:")
print(y.value_counts().sort_index())

# 5. BAGI DATA TANPA STRATIFY
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42
)

print(f"\nData Training: {len(X_train)} baris")
print(f"Data Testing : {len(X_test)} baris")

# 6. LATIH MODEL RANDOM FOREST
model = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    min_samples_split=2,
    random_state=42
)

model.fit(X_train, y_train)
print("\nModel Random Forest berhasil dilatih!")

# 7. PREDIKSI
pred = model.predict(X_test)

# 8. AKURASI
accuracy = accuracy_score(y_test, pred)
print(f"\nAkurasi: {accuracy:.4f} ({accuracy*100:.2f}%)")

# 9. CLASSIFICATION REPORT
print("\n" + "="*60)
print("CLASSIFICATION REPORT")
print("="*60)
print(classification_report(y_test, pred, zero_division=0))

# 10. CONFUSION MATRIX
print("\n" + "="*60)
print("CONFUSION MATRIX")
print("="*60)
cm = confusion_matrix(y_test, pred)
labels = sorted(set(y.unique()) | set(pred))
cm_df = pd.DataFrame(cm, index=labels, columns=labels)
print(cm_df)

# 11. PER HITUNGAN PER KELAS
print("\n" + "="*60)
print("STATISTIK PER KELAS")
print("="*60)

for kelas in sorted(set(y.unique()) | set(pred)):
    total_aktual = (y_test == kelas).sum()
    total_prediksi = (pred == kelas).sum()
    benar_kelas = ((y_test == kelas) & (pred == kelas)).sum()

    if total_aktual > 0:
        recall = benar_kelas / total_aktual
    else:
        recall = 0

    if total_prediksi > 0:
        precision = benar_kelas / total_prediksi
    else:
        precision = 0

    if (precision + recall) > 0:
        f1 = 2 * (precision * recall) / (precision + recall)
    else:
        f1 = 0

    print(f"\nKelas {kelas}:")
    print(f"  Aktual  : {total_aktual} data")
    print(f"  Prediksi: {total_prediksi} data")
    print(f"  Benar   : {benar_kelas} data")
    print(f"  Precision: {precision:.2%}")
    print(f"  Recall   : {recall:.2%}")
    print(f"  F1-Score : {f1:.2%}")

# 12. FEATURE IMPORTANCE
print("\n" + "="*60)
print("FEATURE IMPORTANCE (Top 10)")
print("="*60)

feature_importance = pd.DataFrame({
    'Fitur': fitur_cols,
    'Importance': model.feature_importances_
}).sort_values('Importance', ascending=False)

print(feature_importance.head(10))

# 13. OUTPUT KE KNIME
output = X_test.copy()
output[target_col] = y_test
output["Prediction"] = pred

knio.output_tables[0] = knio.Table.from_pandas(output)

# 14. RINGKASAN
print("\n" + "="*60)
print("RINGKASAN")
print("="*60)
benar = (y_test == pred).sum()
salah = (y_test != pred).sum()
print(f"Total data testing: {len(y_test)}")
print(f"Prediksi benar    : {benar} ({benar/len(y_test)*100:.2f}%)")
print(f"Prediksi salah    : {salah} ({salah/len(y_test)*100:.2f}%)")
print(f"Akurasi           : {accuracy*100:.2f}%")
print("")
print("Selesai!")
```

Dapat mengubah atribut menjadi numerik, kategorikal, string, maupun menentukan atribut sebagai target klasifikasi. Pengaturan tipe data yang benar akan membantu algoritma machine learning menghasilkan model yang lebih sesuai.

Fungsi Edit Domain:
- Mengubah tipe atribut.
- Menentukan target klasifikasi.
- Menyesuaikan format data sebelum proses klasifikasi.

##### 5. k-Nearest Neighbor (kNN)
k-Nearest Neighbor (kNN) digunakan untuk membangun model klasifikasi berdasarkan kedekatan antar data.

Algoritma ini mencari sejumlah tetangga terdekat dari suatu data menggunakan perhitungan jarak. Kelas yang paling banyak muncul pada tetangga tersebut akan menjadi hasil prediksi.

##### 6. Decision Tree
Decision Tree membangun model klasifikasi dalam bentuk pohon keputusan.

Model akan memilih atribut yang paling berpengaruh sebagai percabangan hingga menghasilkan keputusan akhir mengenai kelas suatu data.

##### 7. Naive Bayes
Naive Bayes menggunakan pendekatan probabilitas berdasarkan Teorema Bayes.

Metode ini mengasumsikan bahwa setiap atribut bersifat independen satu sama lain sehingga proses perhitungan menjadi lebih sederhana.

##### 8. Random Forest
Random Forest merupakan algoritma ensemble yang membangun banyak Decision Tree secara bersamaan.

Setiap pohon menghasilkan prediksi, kemudian seluruh prediksi digabungkan menggunakan mekanisme voting untuk menentukan hasil akhir.

##### 9. Test and Score
Test and Score digunakan untuk mengevaluasi performa seluruh model klasifikasi yang telah dibangun.

Pada analisis ini digunakan metode 5-Fold Cross Validation, yaitu teknik validasi yang membagi dataset menjadi lima bagian. Empat bagian digunakan sebagai data latih, sedangkan satu bagian digunakan sebagai data uji. Proses tersebut diulang sebanyak lima kali sehingga setiap data memperoleh kesempatan menjadi data uji.

Hasil evaluasi ditampilkan menggunakan beberapa metrik, yaitu:

- AUC (Area Under Curve) untuk mengukur kemampuan model membedakan kelas.
- CA (Classification Accuracy) untuk mengukur tingkat ketepatan prediksi.
- F1 Score sebagai keseimbangan antara Precision dan Recall.
- Precision untuk mengukur ketepatan prediksi pada kelas tertentu.
- Recall untuk mengukur kemampuan model menemukan seluruh data yang termasuk dalam suatu kelas.
- MCC (Matthews Correlation Coefficient) sebagai ukuran kualitas klasifikasi secara keseluruhan.

Melalui widget ini dapat dilakukan perbandingan performa seluruh algoritma sehingga dapat diketahui model yang memberikan hasil terbaik.

##### 11. ROC Analysis

Kurva ROC menggambarkan hubungan antara True Positive Rate (Sensitivity) dan False Positive Rate (1-Specificity). Semakin mendekati sudut kiri atas, semakin baik kemampuan model dalam melakukan klasifikasi. Selain itu, nilai AUC (Area Under Curve) digunakan sebagai indikator kualitas model, di mana nilai yang semakin mendekati 1 menunjukkan performa klasifikasi yang semakin baik.
