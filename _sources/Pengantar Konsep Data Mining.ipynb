# Pengantar Konsep Data Mining
## 📌Latar Belakang
Sekarang semua aktivitas kita menghasilkan data. Mulai dari belanja online, main media sosial, transaksi bank, sampai sistem akademik di kampus, semuanya menyimpan data. Setiap hari jumlah data terus bertambah dan semakin besar.

Masalahnya, banyak data yang hanya disimpan tanpa benar-benar dimanfaatkan. Padahal di dalam data tersebut sebenarnya ada informasi penting, seperti pola kebiasaan, tren, atau hubungan tertentu yang bisa membantu dalam mengambil keputusan. Nah, di sinilah Penambangan Data (Data Mining) berperan.

## 🌟Definisi
Penambangan data atau data mining adalah proses untuk mencari dan menemukan pola, hubungan, atau informasi penting dari kumpulan data yang berjumlah besar. Data pada dasarnya hanyalah kumpulan angka atau catatan yang belum memiliki arti jika tidak dianalisis. Melalui proses data mining, data mentah tersebut diolah menggunakan teknik tertentu sehingga menghasilkan informasi yang berguna dan dapat membantu dalam pengambilan keputusan.

Istilah “penambangan” digunakan karena prosesnya mirip seperti menambang emas dari dalam tanah. Di antara banyaknya data yang tersimpan, terdapat informasi berharga yang tersembunyi. Dengan bantuan komputer dan metode analisis seperti statistik serta algoritma tertentu, informasi tersebut dapat digali dan dimanfaatkan.

Jadi Data mining adalah teknik untuk mengubah data mentah menjadi informasi yang bermakna melalui proses analisis.

## 🎯Tujuan
Tujuan data mining adalah untuk menemukan pola, hubungan, dan informasi tersembunyi dari kumpulan data yang besar sehingga dapat membantu dalam pengambilan keputusan. Data mining juga digunakan untuk membuat prediksi, mengelompokkan data, serta mendeteksi hal-hal yang tidak biasa dalam data.

## 💡Manfaat
Data mining bermanfaat untuk membantu organisasi atau perusahaan memahami data yang mereka miliki. Dengan data mining, keputusan dapat dibuat lebih cepat, tepat, dan berdasarkan data nyata. Selain itu, data mining dapat meningkatkan efisiensi kerja, mengurangi risiko, serta membantu dalam menyusun strategi yang lebih efektif.

## 🏆Keunggulan
Keunggulan data mining terletak pada kemampuannya mengolah data dalam jumlah besar dengan cepat dan akurat. Data mining juga mampu menemukan pola yang sulit dilihat secara manual serta membantu dalam melakukan prediksi berdasarkan data sebelumnya.

## 📋Tugas Utama
Tugas utama data mining adalah menemukan pola, hubungan, dan informasi penting dari data dalam jumlah besar untuk mendukung pengambilan keputusan. Proses ini mencakup klasifikasi, prediksi, pengelompokan, dan deteksi anomali.

## 📊Prosess
Dalam data mining, terdapat metode standar yang disebut CRISP-DM (Cross-Industry Standard Process for Data Mining). Metode ini merupakan model proses terbuka yang banyak digunakan oleh praktisi data mining untuk mengelola proyek secara sistematis.

CRISP-DM menggambarkan siklus hidup proyek data mining yang terdiri dari beberapa tahapan, mulai dari pemahaman bisnis hingga evaluasi dan implementasi. Setiap tahap memiliki tugas tertentu dan saling terhubung satu sama lain.

### 1️⃣ Memahami Bisnis (Business Understanding)
Business Understanding adalah tahap awal dalam proses data mining yang berfokus pada memahami masalah dan tujuan bisnis sebelum melakukan analisis data. Tahap ini merupakan bagian pertama dari metodologi CRISP-DM. Pada fase ini, analis tidak langsung mengolah data, tetapi memahami kebutuhan organisasi dan menerjemahkannya menjadi tujuan analitis yang jelas dan terukur.

##### Menentukan Tujuan Bisnis (Determine Business Objectives)

Tahap ini berfokus pada identifikasi masalah utama dan penetapan tujuan bisnis yang jelas serta terukur melalui diskusi dengan pihak terkait. Tujuan tersebut menjadi dasar arah proyek dan tolok ukur keberhasilan dari sisi bisnis.

##### Menganalisis Situasi (Assess Situation)

Pada tahap ini dilakukan peninjauan kondisi proyek, termasuk sumber daya, waktu, biaya, risiko, dan keterbatasan yang ada. Analisis ini membantu memastikan proyek layak dan realistis untuk dilaksanakan.

##### Menentukan Tujuan Data Mining (Determine Data Mining Goals)

Tahap ini menerjemahkan tujuan bisnis ke dalam tujuan teknis yang dapat dianalisis menggunakan data, seperti menentukan jenis model dan metode evaluasi. Hal ini memastikan hasil analisis mendukung kebutuhan bisnis.

##### Menyusun Rencana Proyek (Produce Project Plan)

Tahap ini meliputi penyusunan rencana kerja secara sistematis, termasuk jadwal, pembagian tugas, dan tahapan pengerjaan. Perencanaan yang baik membantu proyek berjalan terarah dan efisien.

### 2️⃣ Memahami data (data understanding)
Data Understanding adalah tahap setelah Business Understanding yang bertujuan untuk memahami data yang akan digunakan dalam proyek. Pada tahap ini, tim mulai mengumpulkan data, mengenali struktur dan karakteristiknya, serta mengidentifikasi kualitas data sebelum masuk ke proses pengolahan lebih lanjut.

##### Mengumpulkan Data (Collect Initial Data)

Tahap ini dilakukan dengan mengumpulkan data dari berbagai sumber yang relevan dengan tujuan proyek, seperti database, file, atau sistem internal perusahaan. Proses ini memastikan data yang digunakan sesuai dengan kebutuhan analisis.

##### Mendeskripsikan Data (Describe Data)

Pada tahap ini dilakukan pengenalan terhadap struktur data, seperti jumlah atribut, tipe data, jumlah record, serta gambaran umum isi data. Tujuannya adalah memahami bentuk dan karakteristik dasar dari data yang tersedia.

##### Mengeksplorasi Data (Explore Data)

Tahap ini bertujuan untuk menemukan pola awal, hubungan antar variabel, atau indikasi masalah melalui analisis sederhana seperti statistik deskriptif atau visualisasi. Proses ini membantu memahami perilaku data secara lebih mendalam.

##### Memverifikasi Kualitas Data (Verify Data Quality)

Pada tahap ini dilakukan pemeriksaan terhadap kelengkapan, konsistensi, dan akurasi data, seperti mendeteksi nilai kosong, data duplikat, atau kesalahan input. Tujuannya adalah memastikan data layak digunakan untuk tahap selanjutnya.

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("IRIS.csv")
df_numeric = df.select_dtypes(include=['int64','float64'])

for col in df_numeric.columns:
    print("===== Statistik untuk", col, "=====")
    print("Jumlah Data      :", df[col].count())
    print("Rata-rata        :", df[col].mean())
    print("Nilai Minimum    :", df[col].min())
    print("Q1               :", df[col].quantile(0.25))
    print("Median (Q2)      :", df[col].quantile(0.5))
    print("Q3               :", df[col].quantile(0.75))
    print("Nilai Maximum    :", df[col].max())
    print("Kemencengan      :", round(df[col].skew(), 6))

    mode = stats.mode(df[col], keepdims=True)
    print("Modus            :", mode.mode[0], "dengan jumlah", mode.count[0])

    print("Standar Deviasi  :", round(df[col].std(), 6))
    print("Variansi         :", round(df[col].var(), 6))
    print("\n")
```



### 3️⃣ Persiapan Data (Data Preparation)
Data Preparation adalah tahap yang bertujuan untuk menyiapkan data agar siap digunakan dalam proses pemodelan. Pada tahap ini, data yang sebelumnya telah dipahami akan dibersihkan, dipilih, dan diubah ke dalam format yang sesuai sehingga dapat menghasilkan model yang akurat dan optimal.

##### Memilih Data (Select Data)

Tahap ini dilakukan dengan memilih atribut atau variabel yang relevan dengan tujuan analisis. Tidak semua data digunakan, hanya data yang benar-benar mendukung pencapaian tujuan data mining.

##### Membersihkan Data (Clean Data)

Pada tahap ini dilakukan perbaikan terhadap data yang bermasalah, seperti menangani nilai kosong, menghapus data duplikat, serta memperbaiki kesalahan input. Tujuannya adalah meningkatkan kualitas data agar tidak memengaruhi hasil model.

##### Membangun Data (Construct Data)

Tahap ini melibatkan pembuatan atribut baru atau transformasi variabel untuk meningkatkan informasi yang dapat digunakan model. Contohnya adalah menggabungkan beberapa variabel atau melakukan perhitungan tertentu.

##### Mengintegrasikan Data (Integrate Data)

Pada tahap ini, data dari berbagai sumber digabungkan menjadi satu kesatuan yang konsisten. Proses ini penting jika data berasal dari lebih dari satu sistem atau database.

##### Memformat Data (Format Data)

Tahap terakhir adalah menyesuaikan format data agar sesuai dengan kebutuhan algoritma, seperti melakukan normalisasi, encoding, atau pengubahan tipe data. Setelah tahap ini selesai, data siap digunakan dalam proses pemodelan.

### 4️⃣ Permodelan (Modeling)
Modeling adalah tahap di mana data yang telah dipersiapkan digunakan untuk membangun model analisis sesuai dengan tujuan data mining. Pada tahap ini, berbagai algoritma diterapkan untuk menemukan pola atau membuat prediksi, kemudian hasilnya dievaluasi berdasarkan metrik yang telah ditentukan sebelumnya.

##### Memilih Teknik Pemodelan (Select Modeling Technique)

Tahap ini dilakukan dengan menentukan metode atau algoritma yang sesuai dengan tujuan analisis, seperti klasifikasi, regresi, atau clustering. Pemilihan teknik harus disesuaikan dengan jenis permasalahan dan karakteristik data.

##### Merancang Proses Pengujian (Generate Test Design)

Pada tahap ini ditentukan strategi pengujian model, seperti pembagian data menjadi data latih dan data uji. Tujuannya adalah memastikan model dapat diuji secara objektif dan tidak hanya bekerja baik pada data pelatihan saja.

##### Membangun Model (Build Model)

Tahap ini merupakan proses penerapan algoritma pada data latih untuk menghasilkan model. Dalam proses ini dilakukan penyesuaian parameter agar model dapat bekerja secara optimal.

##### Mengevaluasi Model Secara Teknis (Assess Model)

Pada tahap ini dilakukan pengukuran performa model menggunakan metrik evaluasi seperti akurasi, presisi, atau error rate. Hasil evaluasi ini menentukan apakah model sudah layak untuk dilanjutkan ke tahap berikutnya.

### 5️⃣ Penerapan (Deployment)
Deployment adalah tahap akhir di mana model yang telah dibuat dan dievaluasi diimplementasikan ke dalam lingkungan nyata agar dapat digunakan untuk mendukung pengambilan keputusan bisnis. Tahap ini memastikan bahwa hasil analisis tidak hanya berhenti pada laporan, tetapi benar-benar memberikan manfaat bagi organisasi.

##### Merencanakan Implementasi (Plan Deployment)

Tahap ini dilakukan dengan menyusun strategi penerapan model, termasuk menentukan bagaimana model akan digunakan, siapa yang akan mengoperasikan, serta sistem apa yang akan terlibat. Perencanaan yang baik membantu proses implementasi berjalan lancar.

##### Melakukan Implementasi Model (Deploy Model)

Pada tahap ini, model diintegrasikan ke dalam sistem atau aplikasi yang digunakan perusahaan. Model dapat dijalankan secara otomatis maupun manual sesuai kebutuhan operasional.

##### Monitoring dan Pemeliharaan (Monitor and Maintain)

Setelah diimplementasikan, model perlu dipantau secara berkala untuk memastikan kinerjanya tetap optimal. Jika terjadi perubahan pola data atau penurunan performa, maka model perlu diperbarui atau disesuaikan kembali.
