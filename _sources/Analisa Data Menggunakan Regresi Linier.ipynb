# Analisa Data Menggunakan Regresi Linier

### Regresi linier
merupakan metode statistik dan machine learning yang digunakan untuk memodelkan hubungan antara variabel independen dan variabel dependen dalam bentuk hubungan linear. Metode ini sering digunakan untuk analisis data, prediksi, dan mengetahui pengaruh suatu variabel terhadap variabel lainnya.

### Komponen Model Regresi Linier

1. Model Regresi Linier Sederhana
   $$Y = \beta_0 + \beta_1 X + \epsilon$$

2. Model Regresi Linier Berganda
   $$Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \dots + \beta_n X_n + \epsilon$$

3. Persamaan Hasil Estimasi (Prediksi)
   $$\hat{Y} = b_0 + b_1 X$$

4. Komponen Error / Residual
   $$e = Y - \hat{Y}$$

**Keterangan Simbol:**
* $Y$ = Variabel Terikat (*Dependent Variable*)
* $\hat{Y}$ = Nilai Prediksi Variabel Terikat (*Predicted Value*)
* $X / X_n$ = Variabel Bebas (*Independent Variable*)
* $\beta_0 / b_0$ = Intersep / Titik Potong Sumbu Y (*Intercept*)
* $\beta_1 / b_1$ = Koefisien Regresi / Kemiringan Garis (*Slope*)
* $\epsilon$ = Kesalahan Acak Populasi (*True Error / Disturbance*)
* $e$ = Sisaan Sampel (*Residual*)

#### 1. Tujuan Analisis Regresi Linear Sederhana

Kita ingin mencari **garis lurus terbaik** yang mewakili hubungan antara:

* **Variabel independen (X)** = nilai x dari titik koordinat
* **Variabel dependen (Y)** = nilai y dari titik koordinat

Bentuk umum:

$$Y = \beta_0 + \beta_1X + \varepsilon$$

dimana:

* $\beta_0$ = intercept (titik potong sumbu Y)
* $\beta_1$ = slope (kemiringan garis)
* $\varepsilon$ = error (selisih antara nilai aktual dan prediksi)

Kita akan menghitung $\hat{\beta}_0$ dan $\hat{\beta}_1$ (estimasi koefisien regresi).

#### 3. Perhitungan Manual dengan Rumus Matriks

Kita gunakan rumus:

$$\hat{\beta} = (X^T X)^{-1} X^T Y$$

Keterangan:

| Komponen | Dimensi | Arti |
| :--- | :---: | :--- |
| $X^T$ | $(p + 1) \times n$ | Transpose dari matriks X |
| $X^T X$ | $(p + 1) \times (p + 1)$ | Matriks kovarians (bisa dibalik) |
| $(X^T X)^{-1}$ | $(p + 1) \times (p + 1)$ | Invers dari matriks $X^T X$ |
| $X^T Y$ | $(p + 1) \times 1$ | Vektor korelasi antara X dan Y |
| $\hat{\beta}$ | $(p + 1) \times 1$ | Estimasi koefisien regresi |

##### Bentuk matriks X dan Y

**Matriks X** (dengan kolom pertama berisi angka 1 untuk intercept, kolom kedua berisi nilai x):

$$X = \begin{bmatrix}
1 & 2 \\
1 & 4 \\
1 & 3 \\
1 & 3 \\
1 & 3 \\
1 & 4 \\
1 & 5
\end{bmatrix}$$

**Vektor Y:**

$$Y = \begin{bmatrix}
2 \\
3 \\
5 \\
4 \\
3 \\
5 \\
6
\end{bmatrix}$$

##### Hitung $X^T X$

$$X^T = \begin{bmatrix}
1 & 1 & 1 & 1 & 1 & 1 & 1 \\
2 & 4 & 3 & 3 & 3 & 4 & 5
\end{bmatrix}$$

**Perkalian $X^T X$:**

* Elemen baris 1 kolom 1 = $1^2 + 1^2 + \dots + 1^2 \text{ (7 kali)} = 7$
* Elemen baris 1 kolom 2 = $1 \times 2 + 1 \times 4 + 1 \times 3 + 1 \times 3 + 1 \times 3 + 1 \times 4 + 1 \times 5$
  $= 2 + 4 + 3 + 3 + 3 + 4 + 5 = 24$
* Elemen baris 2 kolom 1 = sama seperti di atas $= 24$
* Elemen baris 2 kolom 2 = $2^2 + 4^2 + 3^2 + 3^2 + 3^2 + 4^2 + 5^2$
  $= 4 + 16 + 9 + 9 + 9 + 16 + 25 = 88$

$$X^T X = \begin{bmatrix}
7 & 24 \\
24 & 88
\end{bmatrix}$$


##### Hitung invers dari $X^T X$

Rumus invers matriks 2x2:

$$\begin{bmatrix}
a & b \\
c & d
\end{bmatrix}^{-1} = \frac{1}{ad - bc} \begin{bmatrix}
d & -b \\
-c & a
\end{bmatrix}$$

dengan $ad - bc = \text{determinan}$.

* $a = 7, b = 24, c = 24, d = 88$
* Determinan $= 7 \times 88 - 24 \times 24 = 616 - 576 = 40$

$$(X^T X)^{-1} = \frac{1}{40} \begin{bmatrix}
88 & -24 \\
-24 & 7
\end{bmatrix} = \begin{bmatrix}
2.2 & -0.6 \\
-0.6 & 0.175
\end{bmatrix}$$

##### Hitung $X^T Y$

$$X^T Y = \begin{bmatrix}
1 & 1 & 1 & 1 & 1 & 1 & 1 \\
2 & 4 & 3 & 3 & 3 & 4 & 5
\end{bmatrix} \begin{bmatrix}
2 \\
3 \\
5 \\
4 \\
3 \\
5 \\
6
\end{bmatrix}$$

* Baris 1 = $2 + 3 + 5 + 4 + 3 + 5 + 6 = 28$
* Baris 2 = $2 \times 2 + 4 \times 3 + 3 \times 5 + 3 \times 4 + 3 \times 3 + 4 \times 5 + 5 \times 6$  
  Hitung satu per satu: $4 + 12 + 15 + 12 + 9 + 20 + 30 = 102$

$$X^T Y = \begin{bmatrix}
28 \\
102
\end{bmatrix}$$

##### Hitung $\hat{\beta} = (X^T X)^{-1}(X^T Y)$

$$\hat{\beta} = \begin{bmatrix}
2.2 & -0.6 \\
-0.6 & 0.175
\end{bmatrix} \begin{bmatrix}
28 \\
102
\end{bmatrix}$$

* Baris 1 ($\hat{\beta}_0$): $2.2 \times 28 + (-0.6) \times 102 = 61.6 - 61.2 = 0.4$
* Baris 2 ($\hat{\beta}_1$): $(-0.6) \times 28 + 0.175 \times 102 = -16.8 + 17.85 = 1.05$

$$\hat{\beta} = \begin{bmatrix}
0.4 \\
1.05
\end{bmatrix}$$

##### Hasil Perhitungan Manual

Persamaan regresi linear:

$$\boxed{Y = 0.4 + 1.05X}$$

* **Intercept ($\beta_0$)** = $0.4$
* **Slope ($\beta_1$)** = $1.05$

Artinya:

* Ketika $X = 0$, nilai prediksi $Y = 0.4$
* Setiap kenaikan $X$ sebesar 1 satuan, $Y$ diperkirakan naik sebesar **1.05 satuan**




#### Implementasi dengan Python & scikit-learn


```python
# Import library
import numpy as np
from sklearn.linear_model import LinearRegression

# Data
X = np.array([[2], [4], [3], [3], [3], [4], [5]])
Y = np.array([2, 3, 5, 4, 3, 5, 6])

# Buat model regresi linear
model = LinearRegression()

model.fit(X, Y)

# Ambil koefisien
intercept = model.intercept_      # beta_0
slope = model.coef_[0]            # beta_1

print(f"Intercept (beta_0): {intercept}")
print(f"Slope (beta_1): {slope}")
print(f"Persamaan regresi: Y = {intercept:.2f} + {slope:.2f} X")
```

Penjelasan kode

| Baris | Penjelasan |
| :--- | :--- |
| `X = np.array([[2],[4],...])` | Bentuk 2D karena sklearn butuh fitur dalam kolom, baris = sampel |
| `Y = np.array([2,3,...])` | Target variable (1D) |
| `LinearRegression()` | Membuat objek model regresi linear |
| `model.fit(X, Y)` | Menghitung koefisien dengan metode least squares |
| `model.intercept_` | Mengambil nilai $\beta_0$ |
| `model.coef_[0]` | Mengambil nilai $\beta_1$ (karena hanya 1 fitur) |

##### 5. Verifikasi Prediksi dengan Data

Coba prediksi untuk $X = 3$:

* Manual: $Y = 0.4 + 1.05 \times 3 = 0.4 + 3.15 = 3.55$
* Kode:

```python
X_new = np.array([[3]])
pred = model.predict(X_new)
print(pred)
```
