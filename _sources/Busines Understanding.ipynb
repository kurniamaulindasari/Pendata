# Business Understanding
## Memahami Bisnis (Business Understanding)
Business Understanding adalah tahap awal dalam proses data mining yang berfokus pada memahami masalah dan tujuan bisnis sebelum melakukan analisis data. Tahap ini merupakan bagian pertama dari metodologi CRISP-DM. Pada fase ini, analis tidak langsung mengolah data, tetapi memahami kebutuhan organisasi dan menerjemahkannya menjadi tujuan analitis yang jelas dan terukur.

- Menentukan Tujuan Bisnis (Determine Business Objectives)

Tahap ini berfokus pada identifikasi masalah utama dan penetapan tujuan bisnis yang jelas serta terukur melalui diskusi dengan pihak terkait. Tujuan tersebut menjadi dasar arah proyek dan tolok ukur keberhasilan dari sisi bisnis.
- Menganalisis Situasi (Assess Situation)

Pada tahap ini dilakukan peninjauan kondisi proyek, termasuk sumber daya, waktu, biaya, risiko, dan keterbatasan yang ada. Analisis ini membantu memastikan proyek layak dan realistis untuk dilaksanakan.

- Menentukan Tujuan Data Mining (Determine Data Mining Goals)

Tahap ini menerjemahkan tujuan bisnis ke dalam tujuan teknis yang dapat dianalisis menggunakan data, seperti menentukan jenis model dan metode evaluasi. Hal ini memastikan hasil analisis mendukung kebutuhan bisnis.

- Menyusun Rencana Proyek (Produce Project Plan)

Tahap ini meliputi penyusunan rencana kerja secara sistematis, termasuk jadwal, pembagian tugas, dan tahapan pengerjaan. Perencanaan yang baik membantu proyek berjalan terarah dan efisien.
