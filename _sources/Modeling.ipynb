# Modeling
## Permodelan (Modeling)
Modeling adalah tahap di mana data yang telah dipersiapkan digunakan untuk membangun model analisis sesuai dengan tujuan data mining. Pada tahap ini, berbagai algoritma diterapkan untuk menemukan pola atau membuat prediksi, kemudian hasilnya dievaluasi berdasarkan metrik yang telah ditentukan sebelumnya.

- Memilih Teknik Pemodelan (Select Modeling Technique)

Tahap ini dilakukan dengan menentukan metode atau algoritma yang sesuai dengan tujuan analisis, seperti klasifikasi, regresi, atau clustering. Pemilihan teknik harus disesuaikan dengan jenis permasalahan dan karakteristik data.

- Merancang Proses Pengujian (Generate Test Design)

Pada tahap ini ditentukan strategi pengujian model, seperti pembagian data menjadi data latih dan data uji. Tujuannya adalah memastikan model dapat diuji secara objektif dan tidak hanya bekerja baik pada data pelatihan saja.

- Membangun Model (Build Model)

Tahap ini merupakan proses penerapan algoritma pada data latih untuk menghasilkan model. Dalam proses ini dilakukan penyesuaian parameter agar model dapat bekerja secara optimal.

- Mengevaluasi Model Secara Teknis (Assess Model)

Pada tahap ini dilakukan pengukuran performa model menggunakan metrik evaluasi seperti akurasi, presisi, atau error rate. Hasil evaluasi ini menentukan apakah model sudah layak untuk dilanjutkan ke tahap berikutnya.
